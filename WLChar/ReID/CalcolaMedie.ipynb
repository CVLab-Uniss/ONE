```python
import os
import csv
import numpy as np

def calcola_medie_cartella(percorso_cartella):
    """
    Legge i file CSV con nome 'reid_iterations_timings_YOLO_cuda_Xvehic.csv' (X=1..9)
    e calcola la media dei valori nella prima colonna (ignorando l'intestazione).
    
    Parametri
    ----------
    percorso_cartella : str
        Il percorso della directory contenente i file CSV.
    
    Ritorna
    -------
    dict
        Dizionario {numero_veicoli: media_colonna1}
    """
    medie = {}
    #device = "cuda"
    device = "cpu"
    
    # ciclo sui file da 1 a 9
    for i in range(1, 10):
        nome_file = f"reid_iterations_timings_YOLO_{device}_{i}vehic.csv"
        percorso_file = os.path.join(percorso_cartella, nome_file)

        # verifica che il file esista
        if not os.path.exists(percorso_file):
            print(f"[AVVISO] File non trovato: {nome_file}")
            continue

        # legge i valori dalla prima colonna (saltando l'intestazione)
        valori = []
        with open(percorso_file, newline='', encoding='utf-8') as csvfile:
            reader = csv.reader(csvfile)
            next(reader)  # salta la prima riga (intestazione)
            for row in reader:
                if row and row[0].strip():  # ignora righe vuote
                    try:
                        valori.append(float(row[0]))
                    except ValueError:
                        print(f"[ATTENZIONE] Valore non numerico in {nome_file}: {row[0]}")

        # calcola la media se ci sono valori
        if valori:
            media = np.mean(valori)
            medie[i] = media
            print(f"File {nome_file}: media = {media:.4f}")
        else:
            print(f"[AVVISO] Nessun dato valido in {nome_file}")

    return medie


#ESEMPIO D’USO:
percorso = "."
risultati = calcola_medie_cartella(percorso)
print(risultati)
```
